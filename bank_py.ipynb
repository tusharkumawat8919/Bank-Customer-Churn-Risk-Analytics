```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('/content/Bank Customer Churn Prediction.csv')

df.head()
```

```python

print(df.info())


print("\nMissing Values:")
print(df.isnull().sum())
```

```python
plt.figure(figsize=(6,4))
sns.countplot(x='churn', data=df, palette='Set2')
plt.title('Bank Customer Churn Count (0 = Stayed, 1 = Left)')
plt.xlabel('Churn Status')
plt.ylabel('Number of Customers')
plt.show()
```

```python
plt.figure(figsize=(10, 5))
sns.histplot(data=df, x='age', hue='churn', multiple='stack', kde=True, palette='coolwarm')
plt.title('Age Distribution by Churn Status')
plt.xlabel('Age')
plt.ylabel('Count')
plt.show()
```

```python
print("Country-wise Churn Percentage:")
print(df.groupby('country')['churn'].mean() * 100)

print("\nGender-wise Churn Percentage:")
print(df.groupby('gender')['churn'].mean() * 100)
```
